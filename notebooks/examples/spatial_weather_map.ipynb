# ECMWF weather map

Explore the latest forecast over the contiguous U.S. and Europe. `weather.plot` reads the grid, UTC valid times, and units from the database frame and selects the lowest-numbered ensemble member in the latest cycle.

Open the exported HTML in a browser to use the time slider, hover readings, and temperature unit controls.

```python
from pathlib import Path
import sys

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src" / "comm_research").is_dir())
sys.path[:0] = [str(ROOT), str(ROOT / "src")]

from comm_research.markets import weather
from infra.database.duckdb_client import DuckDBClient
```

```python
db = DuckDBClient()
df = db.get_data("ecmwf_seasonal")
```

```python
weather_plot = weather.plot(
    df,
    metric="value",
    regions=["us48", "europe"],
    title="ECMWF Hourly Temperature Forecast",
    subtitle="Contiguous U.S. and Europe",
)
weather_plot
```
